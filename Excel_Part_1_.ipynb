**Question 1: What is the difference between 'Paste' and 'Paste Special' in Excel? Briefly explain with examples.**

Answer:-
### **Paste vs Paste Special in Excel**

#### **1. Paste**

* The **normal Paste** option simply pastes *everything* (content, formatting, formulas, comments, etc.) from the copied cell(s) into the new location.
* Shortcut: **Ctrl + V**.

**Example:**
If cell **A1** contains `=5+5` (formula) and is formatted in **bold red font**, copying and pasting it into **B1** will give you:

* Formula `=5+5` (not just the value 10)
* Same **bold red formatting**

So, Paste = exact replica.

---

#### **2. Paste Special**

* **Paste Special** allows you to choose *what exactly* to paste, instead of pasting everything.
* Shortcut: **Ctrl + Alt + V** (opens the Paste Special dialog).
* Options include:

  * **Values** → only the result of formulas.
  * **Formulas** → only formulas (not formatting).
  * **Formats** → only formatting.
  * **Transpose** → switches rows ↔ columns.
  * **Values + Number Formats**, etc.

**Example:**
If cell **A1** contains `=5+5` (formula) and is bold red:

* Paste Special → **Values** into B1 → will paste **10** (result only, no formula or formatting).
* Paste Special → **Formats** → will paste only **bold red style** into B1 (content unchanged).
* Paste Special → **Transpose** → if you copy a row (A1:D1) and paste special with transpose into A2, it will paste as a **column (A2:A5)**.


**Question 2: Describe the functions and usefulness of 'Freeze Panes'** **and 'Split Panes'in Excel.**

Answer:-
### **Freeze Panes vs Split Panes in Excel**

#### **1. Freeze Panes**

* **Function:** Keeps certain rows or columns visible while you scroll through the rest of the worksheet.
* **Usefulness:** Very helpful when working with large datasets where you always want to see **headings (column/row labels)**.
* **Where to find:** Ribbon → **View Tab → Freeze Panes**.

**Examples:**

* **Freeze Top Row** → Always keeps the first row visible (commonly used for headers).
* **Freeze First Column** → Keeps the first column visible when scrolling sideways.
* **Freeze Panes (Custom)** → Freezes rows above and columns to the left of the active cell.

 Example:
If Row 1 has column names (Name, Age, Salary, Dept), using **Freeze Top Row** ensures the headings stay visible while scrolling down through thousands of rows.

---

#### **2. Split Panes**

* **Function:** Splits the worksheet window into two or four independent scrollable sections.
* **Usefulness:** Lets you view and compare different parts of a worksheet at the same time.
* **Where to find:** Ribbon → **View Tab → Split**.

**Examples:**

* If you split horizontally at Row 10 → You can scroll rows 1–10 separately from the rest of the sheet.
* If you split vertically at Column C → You can scroll columns A–C separately from D onward.
* You can also split both ways to create **four quadrants** of the worksheet.

 **Example**:
If you have employee data in thousands of rows, you can split at row 1 and column A. Now you can scroll through the data while still keeping headings (row 1) and employee IDs (column A) in view.


**Question 3: Explain the difference between inserting a new row and** **inserting a new column in Excel. Can you insert multiple rows** or columns at once?

Answer:-
### **Difference Between Inserting a Row and a Column**

#### **1. Inserting a New Row**

* When you insert a new row, Excel **shifts the existing rows downward**.
* The new row is added **above the selected row**.
* It will have the same formatting as the row above it.

**Example:**
If you have data in Rows 1–5 and you insert a row at Row 3 → The original Row 3 (and everything below) becomes Row 4, and a blank row appears at Row 3.

---

#### **2. Inserting a New Column**

* When you insert a new column, Excel **shifts the existing columns to the right**.
* The new column is added **to the left of the selected column**.
* It will have the same formatting as the column on its left.

**Example:**
If you have data in Columns A–D and you insert a column at Column B → The original Column B becomes Column C, and a blank column appears at Column B.

---

### **Can You Insert Multiple Rows or Columns at Once?**

 **Yes**

* Select the number of rows/columns you want to insert → Right-click → **Insert**.
* Excel will insert the same number of blank rows or columns.

**Examples:**

* To insert **3 new rows** → Select **3 rows** (say Rows 4–6), then right-click → Insert → Excel will insert 3 blank rows starting from Row 4.
* To insert **2 new columns** → Select **2 columns** (say Columns C and D), then right-click → Insert → Excel will insert 2 blank columns starting from Column C.

Question 4: What are logical functions in Excel? Provide examples of at least two logical functions and their applications.

Answer:-
### **Logical Functions in Excel**

* **Definition:** Logical functions in Excel are used to perform **comparisons** or **tests** that return either **TRUE** or **FALSE**, or take action based on conditions.
* **Purpose:** They help in **decision-making**, data validation, and conditional calculations.

---

### **Examples of Logical Functions**

#### **1. IF Function**

* **Syntax:**
  `=IF(logical_test, value_if_true, value_if_false)`
* **Use:** Checks a condition and returns one value if the condition is **TRUE** and another if it is **FALSE**.

**Example:**
`=IF(B2>=50,"Pass","Fail")`

* If the value in **B2** is 50 or more → returns **"Pass"**
* Otherwise → returns **"Fail"**

 **Application:** Useful in grading systems, eligibility checks, sales target analysis, etc.

---

#### **2. AND Function**

* **Syntax:**
  `=AND(condition1, condition2, …)`
* **Use:** Returns **TRUE** only if **all conditions** are TRUE, otherwise FALSE.

**Example:**
`=AND(B2>=50, C2="Yes")`

* Returns **TRUE** only if B2 is **50 or more** **and** C2 contains "Yes".

 **Application:** Can be combined with IF to check multiple conditions at once.
Example: `=IF(AND(B2>=50,C2="Yes"),"Eligible","Not Eligible")`

---

#### **3. OR Function** (bonus)

* **Syntax:**
  `=OR(condition1, condition2, …)`
* **Use:** Returns **TRUE** if **any one condition** is TRUE.

**Example:**
`=OR(B2>=50, C2="Yes")`

* Returns **TRUE** if either condition is satisfied.

 **Application:** Useful when multiple alternate conditions can qualify for a decision.



Question 5: Discuss the purpose of 'XLOOKUP' and how it differs from the traditional 'VLOOKUP' function.

Answer:-

### **Purpose of XLOOKUP**

* **XLOOKUP** is used to **search for a value in a range (or array) and return a corresponding value** from another range.
* It is the modern replacement for **VLOOKUP** and **HLOOKUP**, introduced in Excel 365 and Excel 2019.
* It is **more flexible, powerful, and easier to use** than traditional lookup functions.

**Basic Syntax:**

```
=XLOOKUP(lookup_value, lookup_array, return_array, [if_not_found], [match_mode], [search_mode])
```

---

### **How XLOOKUP Differs from VLOOKUP**

| Feature                   | **VLOOKUP**                                                       | **XLOOKUP**                                                          |
| ------------------------- | ----------------------------------------------------------------- | -------------------------------------------------------------------- |
| **Direction**             | Searches **vertically** only (first column of a table).           | Searches **both vertically and horizontally**.                       |
| **Column Restriction**    | Return column must be **to the right** of lookup column.          | Can return values from **any column (left or right)**.               |
| **Exact Match (default)** | Defaults to **approximate match** (unless specified).             | Defaults to **exact match** (safer and more intuitive).              |
| **Syntax Simplicity**     | Needs column index number (e.g., `=VLOOKUP(101,A2:D10,3,FALSE)`). | Directly selects return range (e.g., `=XLOOKUP(101,A2:A10,C2:C10)`). |
| **Error Handling**        | Returns **#N/A** if value not found (requires IFERROR to handle). | Has a built-in `[if_not_found]` argument (e.g., `"Not Found"`).      |
| **Performance**           | Slower on large datasets.                                         | Faster and more efficient.                                           |

---

### **Examples**

#### **1. VLOOKUP Example**

```
=VLOOKUP(101, A2:D10, 3, FALSE)
```

* Looks for **101** in the first column of `A2:D10`.
* Returns the value from the **3rd column** of the range.
* Problem: If you insert/delete a column, the column index breaks.

---

#### **2. XLOOKUP Example**

```
=XLOOKUP(101, A2:A10, C2:C10, "Not Found")
```

* Looks for **101** in `A2:A10`.
* Returns the value from `C2:C10` (directly specified, no index number).
* If not found → returns `"Not Found"`.

## **Detecting and Handling Missing Data in Excel**

### **1. Using Go To Special**

* **Purpose:** Quickly select all blank cells in a dataset.
* **Steps:**

  1. Select the dataset range.
  2. Go to **Home → Find & Select → Go To Special → Blanks → OK**.
  3. All blank cells are highlighted.
* **Usefulness:** Once selected, you can:

  * Fill missing values manually
  * Use formulas to fill in values (e.g., `0`, `"Unknown"`, or average of column)


### **2. Using ISBLANK Function**

* **Purpose:** Check if a particular cell is blank. Returns **TRUE** if blank, otherwise **FALSE**.

* **Syntax:**

  ```
  =ISBLANK(cell)
  ```

* **Example:**

  ```
  =ISBLANK(B2)
  ```

  * Returns **TRUE** if B2 is empty.



---

### **3. Using COUNTBLANK Function**

* **Purpose:** Count the total number of blank cells in a range.

* **Syntax:**

  ```
  =COUNTBLANK(range)
  ```

* **Example:**

  ```
  =COUNTBLANK(B2:B6)
  ```

  * Returns **2**, meaning there are 2 missing ages in the column.

* **Usefulness:** Quickly quantify missing data before deciding how to handle it.

---

### **Handling Missing Data**

Once detected, you can handle blanks in several ways:

1. **Fill with a constant:** e.g., `0` or `"Unknown"`
2. **Fill with average/median of column:** e.g., `=AVERAGE(B2:B6)`
3. **Delete the row:** if the missing data makes the row unusable
